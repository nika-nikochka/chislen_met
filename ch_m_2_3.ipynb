```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

### Задание 1

```python
def phi(x):
    return -0.5*x + 1.5

x = np.linspace(0, 2, 100)
plt.plot(x, phi(x), 'b-', linewidth=2, label='φ(x) = -0.5x + 1.5')
plt.plot(x, x, 'r--', linewidth=2, label='y = x')
plt.grid(True, alpha=0.3)

x0 = 0.2
x_points = [x0]
n_iterations = 8

for i in range(n_iterations):
    x_next = phi(x_points[-1])
    x_points.append(x_next)
    
    plt.plot([x_points[i], x_points[i]], [x_points[i], x_points[i+1]], 'g-', linewidth=1.5)
    plt.plot([x_points[i], x_points[i+1]], [x_points[i+1], x_points[i+1]], 'g-', linewidth=1.5)
    plt.plot(x_points[i], phi(x_points[i]), 'ro', markersize=6)

plt.title(f'Сходящийся итерационный процесс\nφ(x) = -0.5x + 1.5, φ\'(x) = -0.5')
plt.xlabel('x'), plt.ylabel('y')
plt.legend()
plt.axis([0, 2, 0, 2])
plt.grid(True, alpha=0.3)
plt.gca().set_aspect('equal')
plt.tight_layout()
plt.show()
```

### Задание 2

```python
print("Решение уравнения x·sin(x) - 1 = 0")
print("Метод простой итерации, точность 10^-5")
print()

def phi(x):
    return 1 / np.sin(x)

x = 1

for n in range(1, 20):
    x_new = phi(x)
    print(f"Итерация {n}: x = {x_new:.8f}")
    
    if abs(x_new - x) < 0.00001:
        print()
        print(f"Приближённое значение корня: {x_new:.8f}")
        print(f"Проверка: x * sin(x) - 1 = {x_new * np.sin(x_new) - 1:.8f}")
        break
    x = x_new
```

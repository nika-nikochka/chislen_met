```python
''' Задание E '''

class Parrot:
    def __init__(self, text = 'Кукареку!'):
        self.phrases = [text]
    
    def say(self, n = 1):
        for _ in range(n):
            print(random.choice(self.phrases), end=" ")
        print()
       
    def learn(self, text):
        self.phrases.append(text)

p = Parrot("Гав!")
p.say()
p.learn("Мяу!")
p.say()
p.say(3)
```

```python
''' Задание C '''

class LampRow:

    def __init__(self, n=8):
        self.__n = n
        self.__state = "0" * n
    
    @property
    def state(self):
        return self.__state
    
    @state.setter
    def state(self, new_state):
        if len(new_state) == self.__n and all(c in '012' for c in new_state):
            self.__state = new_state
        else:
            self.__state = "0" * self.__n
    
    def show(self):
        res = ""
        for lamp in self.__state:
            if lamp == '0':
                res += "-"
            elif lamp == '1':
                res += "*"
            else:
                res += "o"
        print(res)


lamps = LampRow(6)
lamps.show()
lamps.state = "102102"
print(lamps.state)
lamps.show()

print("\nПример с ошибкой:")
lamps.state = "10201010"
print(lamps.state) 
lamps.show()
```

```python
''' Задание D '''

class LampRow:
    
    def __init__(self, n = 8):
        self.__n = n
        self.__state = 0
    
    @property
    def state(self):
        if self.__state == 0:
            return "0" * self.__n
        
        res = ""
        temp = self.__state
        for _ in range(self.__n):
            dig = temp % 3
            res = str(dig) + res
            temp //= 3
        return res
    
    @state.setter
    def state(self, new_state):
        if len(new_state) == self.__n and all(c in '012' for c in new_state):
            value = 0
            for i, dig in enumerate(new_state):
                value = value * 3 + int(dig)
            self.__state = value
        else:
            self.__state = 0
    
    def show(self):
        state_str = self.state
        res = ""
        for lamp in state_str:
            if lamp == '0':
                res += "-"
            elif lamp == '1':
                res += "*"
            else:
                res += "o"
        print(res)


lamps = LampRow(6)
lamps.show()
lamps.state = "102102"
print(lamps.state)
lamps.show()

print("\nПример с ошибкой:")
lamps.state = "10201010"
print(lamps.state)
lamps.show()
```

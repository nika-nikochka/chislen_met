```python
import math
```

```python
def bernoulli_exact(n, m, p):
    if n <= 1000:
        c = math.comb(n, m)
        return c * (p**m) * ((1-p)**(n-m))
    else:
        log_с = (math.log(math.comb(n, m)) + m * math.log(p) + (n - m) * math.log(1 - p))
        return math.exp(log_с)
```

### задание 1

```python
n = 100000
p_zash = 0.94
m = 5800

p_ver = bernoulli_exact(n, m, 1 - p_zash)
print(f"n = {n}, m = {m}, p = {p_zash}")
print(f"P({m}) = {p_ver:.6f}")
```
